```python
df = df.drop(columns=['_input', '_widgetName', '_source', '_pageUrl', '_resultNumber'])
```

```python
df['metro'] = df.groupby('city')['metro'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown'))
```

```python
df.dtypes
```

```python
df['revenue'] = df['revenue'].astype(str).str.replace(',', '').astype(float)
```

```python
df['revenue_per_employee'] = df['revenue'] / df['workers']
```

```python
df['growth_category'] = pd.cut(df['growth'],
                              bins=[0, 1000, 10000, 50000, 200000],
                              labels=['Low', 'Medium', 'High', 'Very High'])
```

```python
df['industry'] = df['industry'].str.strip()
df['state_s'] = df['state_s'].str.upper()
```
